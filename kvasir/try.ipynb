```python
import os
import pandas as pd
```

```python
df = pd.DataFrame()
labels = {}
ds_path = '../../Datasets/kvasir-dataset-v2'
for idx, dir in enumerate(os.listdir(ds_path)):
    labels[dir] = idx
    temp = {'id_code': [], 'diagnosis': []}
    for img in os.listdir(os.path.join(ds_path, dir)):
        if not img.endswith('jpg'):
            continue
        temp['id_code'].append(dir+'/'+img[:-4])
        temp['diagnosis']=idx
    temp = pd.DataFrame(temp)
    df = pd.concat([df, temp], ignore_index=True)

print(labels)
```

```python
from sklearn.model_selection import train_test_split

train, test = train_test_split(df, test_size=0.2, random_state=42)
```

```python
train.to_csv('split1_train.csv', index=False)
test.to_csv('split1_test.csv', index=False)
```
